```python
import torch

tensor_1d = torch.tensor([1, 2, 3])
print("1D Tensor (Vector):")
print(tensor_1d)
print()

square_tensor=tensor_1d**2

print("squared tensor")

print(square_tensor)

print("multiplication of the tensor")

multiplication=tensor_1d * 2
print(multiplication)
```

```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from PIL import Image
from torchvision.transforms import ToPILImage


image_path="/home/iid/agenticai/demo-project-pytorch-tutorial/image/girl.jpg"
image=Image.open(image_path)

plt.imshow(image)
plt.title("Loaded Image")
plt.axis('on')
plt.show()

image_transform=transforms.ToTensor()
tensor_image=image_transform(image)
plt.imshow(tensor_image.permute(1,2,0))
plt.title("Loaded Image")
plt.axis('on')
plt.show()

print("converting the tensor into image")

to_pil = ToPILImage()

recovered_image = to_pil(tensor_image)

plt.imshow(recovered_image)
plt.title("Tensor → Image")
plt.axis('off')
plt.show()


```
